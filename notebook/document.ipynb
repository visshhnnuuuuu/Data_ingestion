```python
### create the simple text file
import os
os.makedirs("../data/text_files", exist_ok=True)
```

```python
sample_text = {"../data/text_files/sample.txt":
 """
 Python is the global standard and dominant programming language for Artificial Intelligence (AI) and Machine Learning (ML). Holding a significant majority in developer preference, it powers everything from foundational machine learning algorithms to modern Large Language Model (LLM) agent frameworks and Generative AI applications.
  Python's extensive ecosystem of libraries, frameworks, and tools makes it the go-to choice for AI/ML development, enabling rapid prototyping, efficient data manipulation, and seamless integration with cutting-edge technologies. Its versatility and ease of use have solidified Python's position as the preferred language for AI/ML practitioners worldwide.
 """}
```

```python
for filepath, content in sample_text.items():
    with open(filepath, 'w', encoding='utf-8') as f:
        f.write(content)
    print(f"Created sample text file: {filepath}")
```

```python
###textloader
from langchain.document_loaders import TextLoader
```

```python
loader = TextLoader("../data/text_files/sample.txt")
document=loader.load()
print(document)
```
